```python
def get_model_memory_footprint(model):
    """Calculates and returns the model's memory footprint in MB."""
    mem_params = sum(param.nelement() * param.element_size() for param in model.parameters())
    mem_bufs = sum(buf.nelement() * buf.element_size() for buf in model.buffers())
    total_mem_bytes = mem_params + mem_bufs
    return total_mem_bytes / (1024 ** 2) # Convert bytes to MB
```

```python
def load_model(MODEL_NAME, DEVICE, quantization_config=None):
    """TODO: Implement the logic for loading a tokenizer and model."""
    print("Loading tokenizer...")
    tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME,
                                              padding_side="left", 
                                              trust_remote_code= True
                                             )
    print("Tokenizer loaded successfully.")
    print("\nLoading model...")
    if quantization_config is None:
        model = AutoModelForCausalLM.from_pretrained(
            MODEL_NAME,
            torch_dtype=torch.float16,
            device_map= "balanced",
            trust_remote_code =True
            ).to(DEVICE)
    else:
        model = AutoModelForCausalLM.from_pretrained(
            MODEL_NAME,
            quantization_config=quantization_config,
            device_map= "auto",
            trust_remote_code =True
            ).to(DEVICE)
    print("Model loaded successfully and moved to device.")
    memory_mb = get_model_memory_footprint(model)
    print(f"Memory Footprint: {memory_mb:.2f} MB")
  
    model.config.eos_token_id = tokenizer.eos_token_id
    tokenizer.pad_token = tokenizer.eos_token
    tokenizer.pad_token_id = tokenizer.eos_token_id
    model.config.pad_token_id = model.config.eos_token_id
    model.eval()
    return tokenizer, model
```

```python
def generate_config(MAX_NEW_TOKENS, temperature=0.6, top_p=0.5, top_k=3, do_sample=True, use_cache=False):
    generation_config = {
        "max_new_tokens": MAX_NEW_TOKENS,
        "temperature": temperature,           # Controls randomness (lower = more deterministic)
        "top_p": top_p,                 # Nucleus sampling for diversity
        "top_k": top_k,                  # Limits candidates per step
        "do_sample": do_sample,            # Enables stochastic sampling
        # Repetition control
        "repetition_penalty": 1.2,    # Penalizes repeated tokens
        # Efficiency / inference tuning
        "use_cache": use_cache,            # Cache KV states for faster decoding
        # Optional advanced knobs
        "min_new_tokens": 10,         # Guarantees some output
        "eos_token_id": tokenizer.eos_token_id,
        "pad_token_id": tokenizer.pad_token_id,

        }
    return generation_config
```

```python
def apply_chat_temp(example, generate_text=False):
    if generate_text:
        toks = tokenizer.apply_chat_template(example['text_gen'],tokenize=True, add_generation_prompt=True)
    else:
        toks = tokenizer.apply_chat_template(example['text'],tokenize=True)
    return toks
```

```python
def cuda_warmup(model, toks, device, num_runs=5):
    for runs in range(num_runs):
        with torch.inference_mode():
            _ = model.generate(input_ids=toks, max_new_tokens=15)
        if device == "cuda":
            torch.cuda.synchronize()
    print("WarmUp runs completed")
```

```python
import string

def remove_punctuation(text: str) -> str:
    # Create a translation table that maps punctuation to None
    translator = str.maketrans('', '', string.punctuation)
    return text.translate(translator)
```

```python
def generate_headline(model, tokenizer, toks, device, generation_args):
    """TODO: Implement the headline generation and latency measurement logic."""
    t0 = time.perf_counter()
    with torch.inference_mode():
            outputs = model.generate(**toks, **generation_args)
    if device == 'cuda':
        torch.cuda.synchronize()
    t1 = time.perf_counter()
    latency = t1 - t0
    gen_len = outputs.shape[1] - toks["input_ids"].shape[1]
    peak_mem_mib = torch.cuda.max_memory_allocated(device) / 1024**2
    throughput = gen_len / latency if latency > 0 else float("nan")

    outputs_text = tokenizer.decode(outputs)
    outputs_text = remove_punctuation(outputs_text[0].split('<|start_header_id|>assistant<|end_header_id|>')[-1].split('<|eot_id|>')[0].strip())
    #print(outputs_text)
    peak_mem_mib = torch.cuda.max_memory_allocated(device) / 1024**2
    del model
    return latency, throughput, peak_mem_mib, outputs_text
```

```python
def report_metrics(tok_ds, model, tokenizer, DEVICE, generation_args, exp, df):
    """TODO: Implement the logic for calculating and reporting all performance metrics."""
    avg_latency, avg_throughput, avg_peak_mib, rouge_score = evaluate_model(tok_ds, model, tokenizer, DEVICE, generation_args)
    new_df = pd.DataFrame({"EXP NAME": [exp], "Mean Latency (s)": [avg_latency], "Throughput (tokens/s)": [avg_throughput], "Avg Peak Mem": [avg_peak_mib], "ROUGE-1 Score": [rouge_score]}) 
    df = pd.concat([df, new_df], axis=0)
    return df
```

```python
def evaluate_model(dataset, model, tokenizer, device, generation_args, n=4):
    """TODO: Implement the model evaluation loop."""
    
    sample = random.sample(range(1000), 1)[0]
    warmup_example = dataset['train'][sample]
    toks = apply_chat_temp(warmup_example, generate_text=True)

    toks['input_ids'] = torch.tensor(toks['input_ids']).unsqueeze(0).to(device)
    toks['attention_mask'] = torch.tensor(toks['attention_mask']).unsqueeze(0).to(device)
    
    model.eval()
    cuda_warmup(model, toks['input_ids'], device)
    
    torch.cuda.reset_peak_memory_stats(device)
    samples = 100
    latencies = []
    throughputs = []
    peak_mem = []
    outputs = []
    references = []
    for i in range(n):
        for j in range(samples):
            sample = dataset['train'][j]
            references.append(sample['headline'])
            toks = apply_chat_temp(sample, generate_text=True)
            toks['input_ids'] = torch.tensor(toks['input_ids']).unsqueeze(0).to(model.device)
            toks['attention_mask'] = torch.tensor(toks['attention_mask']).unsqueeze(0).to(model.device)
            latency, throughput, peak_mem_mib, outputs_text = generate_headline(model, tokenizer, toks, device, generation_args)
            latencies.append(latency)
            throughputs.append(throughput)
            peak_mem.append(peak_mem_mib)
            outputs.append(outputs_text)
    
    rouge = evaluate.load("rouge")
    rouge_score = rouge.compute(predictions=outputs, references=references)['rouge1']
    avg_latency = sum(latencies) / len(latencies)
    avg_throughput = sum(throughputs) / len(throughputs)
    avg_peak_mib = sum(peak_mem)/len(peak_mem)
    print(avg_latency, avg_throughput, avg_peak_mib, rouge_score)
    return avg_latency, avg_throughput, avg_peak_mib, rouge_score
```

```python
import copy
import torch.nn as nn
def prune_model_weights(model, amount=0.3):
    """TODO: Applies L1 unstructured pruning to the linear layers of a model."""
    pruned_model = copy.deepcopy(model)
    for name, module in pruned_model.named_modules():
        if isinstance(module, nn.Linear):
            prune.l1_unstructured(module, name="weight", amount=amount)
            prune.remove(module, "weight")
    print("Model pruned successfully")
    return pruned_model
```

```python
def generate_headline_batch(model, tokenizer, device, generation_args, data_loader, exp, df):
    """TODO: Implement the headline generation and latency measurement logic."""
    for batch in data_loader:
        warmup_example = batch["input_ids"].to(device)
        model.eval()
        cuda_warmup(model, warmup_example, device)
        break
        
    torch.cuda.reset_peak_memory_stats(device)
    
    latencies_per_batch = []
    throughputs = []
    peak_mem = []
    batch_outputs = []
    references = []
    for step, batch in enumerate(data_loader):
        print(f"Batch {step}")
        #print("input_ids.shape:", batch["input_ids"].shape)
        #print("headline_ids.shape:", batch["headline_ids"].shape)       
        t0 = time.perf_counter()
        batch["input_ids"] = batch["input_ids"].to(model.device)
        batch['attention_mask'] = batch['attention_mask'].to(model.device)
        with torch.inference_mode():
            outputs = model.generate(input_ids=batch["input_ids"], attention_mask=batch['attention_mask'], **generation_args)
        if device == 'cuda':
            torch.cuda.synchronize()
        t1 = time.perf_counter()
        
        latency_per_batch = (t1 - t0)
        latencies_per_batch.append(latency_per_batch)

        peak_mem_mib = torch.cuda.max_memory_allocated(device) / 1024**2
        peak_mem.append(peak_mem_mib)
        
        inputs_decode = tokenizer.batch_decode(batch["input_ids"], skip_special_tokens=True)
        inputs_toks_len_list = [len(tokenizer.encode(inp)) for inp in inputs_decode]
        
        outputs_decode = tokenizer.batch_decode(outputs, skip_special_tokens=True)
        outputs_text = [remove_punctuation(out.split('assistant\n\n')[-1]) for out in outputs_decode]
        batch_outputs.extend(outputs_text)
        
        outputs_toks_len_list = [len(tokenizer.encode(out)) for out in outputs_decode]
        
        num_toks_gen = sum(np.array(outputs_toks_len_list) - np.array(inputs_toks_len_list))
        throughput = num_toks_gen / latency_per_batch if latency_per_batch > 0 else float("nan")
        throughputs.append(throughput)
        
        headline_decode = tokenizer.batch_decode(batch["headline_ids"], skip_special_tokens=True)
        references.extend(headline_decode)


    rouge = evaluate.load("rouge")
    rouge_score = rouge.compute(predictions=batch_outputs, references=references)['rouge1']
    
    avg_latency_per_batch = sum(latencies_per_batch) / len(latencies_per_batch)
    avg_latency_per_batch_item = sum(latencies_per_batch) / len(data_loader.dataset)
    
    avg_throughput = sum(throughputs) / len(throughputs)
    avg_peak_mib = sum(peak_mem)/len(peak_mem)
    
    print(avg_latency_per_batch, avg_latency_per_batch_item, avg_throughput, avg_peak_mib, rouge_score)
    
    new_df = pd.DataFrame({"EXP NAME": [exp], "Mean Latency (s)": [f"{round(avg_latency_per_batch, 2)}/{round(avg_latency_per_batch_item, 2)}"], "Throughput (tokens/s)": [round(avg_throughput,2)], "Avg Peak Mem": [round(avg_peak_mib, 2)], "ROUGE-1 Score": [round(rouge_score, 2)]}) 
    df = pd.concat([df, new_df], axis=0)
    
    return df
```

# 6. Distributed Inference (Multi-GPU)

**Your Task:** If you have multiple GPUs, you can split the model across them to reduce the memory burden on a single GPU and potentially improve latency. We will explore two common techniques: Tensor Parallelism and Pipeline Parallelism.

*Note: This section requires a multi-GPU environment.*

### Tensor Parallelism
Tensor parallelism splits individual model layers (the tensors) across multiple GPUs. Operations like matrix multiplications are executed in parallel on different GPUs, and the results are aggregated. This is highly effective for reducing the memory footprint of very large layers. The `accelerate` library can handle this automatically via `device_map="auto"`.

### Pipeline Parallelism
Pipeline parallelism assigns entire layers or blocks of layers to different GPUs, creating a sequence or "pipeline" that the data flows through. For example, layers 1-10 run on GPU 0, layers 11-20 run on GPU 1, and so on. This is useful for very deep models where even a single layer might be too large for one GPU after tensor parallelism.

```python
def speculate(new_inp, draft_model, target_model, device, generation_args_draft, generation_args_target):
    
    orig_toks_len = new_inp.shape[1]
    #print(f'Sepculate: {orig_toks_len}')
    with torch.inference_mode():
        out_draft = draft_model.generate(input_ids=new_inp, **generation_args_draft)
    if device == 'cuda':
        torch.cuda.synchronize()
    out_draft_len = out_draft.shape[1]
    delta_toks  = out_draft_len - orig_toks_len
            
    new_toks = new_inp[0].tolist() + out_draft[0][-delta_toks:].tolist()
    delta_draft_toks = new_toks[-delta_toks:]
    new_toks = torch.tensor(new_toks).unsqueeze(0).to(device)
    with torch.inference_mode():
        out_target = target_model.generate(input_ids=new_toks,  **generation_args_target)
    if device == 'cuda':
        torch.cuda.synchronize()
    delta_target_toks = out_target[0].tolist()[-(delta_toks+1):-1]
    return delta_draft_toks, delta_target_toks
```

```python
def match(delta_draft_toks, delta_target_toks, delta_agreed_toks):
    for i, d in enumerate(delta_draft_toks):
        if d == delta_target_toks[i]:
            delta_agreed_toks.append(d)
        else:
            break
    return delta_agreed_toks
```

```python
def generate_spec_headline(draft_model, target_model, tokenizer, toks, device, generation_args, generation_args_draft, generation_args_target, delta_agreed_toks):
    """TODO: Implement the headline generation and latency measurement logic."""

    t0 = time.perf_counter()
    delta_draft_toks, delta_target_toks =  speculate(toks, draft_model, target_model, device, generation_args_draft, generation_args_target)
    delta_agreed_toks = match(delta_draft_toks, delta_target_toks, delta_agreed_toks)
    
    while delta_agreed_toks[-1] != tokenizer.eos_token_id:
        if (len(delta_agreed_toks) < generation_args['max_new_tokens']):
            new_inp = torch.tensor(toks[0].tolist() + delta_agreed_toks).unsqueeze(0).to(device)
            delta_draft_toks, delta_target_toks = speculate(new_inp, draft_model, target_model, device, generation_args_draft, generation_args_target)
            delta_agreed_toks = match(delta_draft_toks, delta_target_toks, delta_agreed_toks)
            #print(f'delta agreed toks {delta_agreed_toks}')
        else:
            break
    ################################################################
    t1 = time.perf_counter()
    #print(tokenizer.decode(toks[0].tolist()))
    #print(tokenizer.decode(delta_agreed_toks))
    final_output = toks[0].tolist() + delta_agreed_toks
    #print(tokenizer.decode(final_output))
    latency = t1 - t0
    gen_len = len(delta_agreed_toks)
    peak_mem_mib = torch.cuda.max_memory_allocated(device) / 1024**2
    throughput = gen_len / latency if latency > 0 else float("nan")
    outputs_text = tokenizer.decode(final_output)
    #print(outputs_text)
    #print("#################################################################")
    outputs_text = remove_punctuation(outputs_text.split('<|start_header_id|>assistant<|end_header_id|>')[-1].split('<|eot_id|>')[0].strip())
    #print(outputs_text)
    return latency, throughput, peak_mem_mib, outputs_text
```

```python
def eval_spec(dataset, draft_model, target_model, tokenizer, device, generation_args, generation_args_draft, generation_args_target, n=2):
    sample = random.sample(range(1000), 1)[0]
    warmup_example = dataset['train'][sample]
    toks = apply_chat_temp(warmup_example, generate_text=True)
    toks['input_ids'] = torch.tensor(toks['input_ids']).unsqueeze(0).to(device)
    toks['attention_mask'] = torch.tensor(toks['attention_mask']).unsqueeze(0).to(device)
    cuda_warmup(draft_model, toks['input_ids'], device)
    cuda_warmup(target_model, toks['input_ids'], device)

    torch.cuda.reset_peak_memory_stats(device)
    samples = 100
    latencies = []
    throughputs = []
    peak_mem = []
    outputs = []
    references = []
    delta_agreed_toks = []

    for i in range(n):
        for j in range(samples):
            delta_agreed_toks = []
            sample = dataset['train'][j]
            references.append(sample['headline'])
            toks = apply_chat_temp(sample, generate_text=True)
            toks['input_ids'] = torch.tensor(toks['input_ids']).unsqueeze(0).to(device)
            orig_toks_len = toks['input_ids'].shape[1]
            latency, throughput, peak_mem_mib, outputs_text = generate_spec_headline(draft_model, target_model, tokenizer, toks['input_ids'], device, generation_args, generation_args_draft, generation_args_target, delta_agreed_toks)
            latencies.append(latency)
            throughputs.append(throughput)
            peak_mem.append(peak_mem_mib)
            outputs.append(outputs_text)
    rouge = evaluate.load("rouge")
    rouge_score = rouge.compute(predictions=outputs, references=references)['rouge1']
    avg_latency = sum(latencies) / len(latencies)
    avg_throughput = sum(throughputs) / len(throughputs)
    avg_peak_mib = sum(peak_mem)/len(peak_mem)
    print(avg_latency, avg_throughput, avg_peak_mib, rouge_score)
    #print(outputs)
    #print("####################################")
    #print(references)
    return avg_latency, avg_throughput, avg_peak_mib, rouge_score   
```

```python
def report_spec_metrics(tok_ds,  draft_model, target_model, tokenizer, device, generation_args, generation_args_draft, generation_args_target, exp, df):
    """TODO: Implement the logic for calculating and reporting all performance metrics."""
    avg_latency, avg_throughput, avg_peak_mib, rouge_score = eval_spec(tok_ds, draft_model, target_model, tokenizer, device, generation_args, generation_args_draft, generation_args_target, n=4)
    new_df = pd.DataFrame({"EXP NAME": [exp], "Mean Latency (s)": [avg_latency], "Throughput (tokens/s)": [avg_throughput], "Avg Peak Mem": [avg_peak_mib], "ROUGE-1 Score": [rouge_score]}) 
    df = pd.concat([df, new_df], axis=0)
    return df
```
